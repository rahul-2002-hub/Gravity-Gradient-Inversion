```python
# =====================================================================
# REAL-DATA CONVERGENCE CURVES: Gz, Txy, Tzx, Tzy
# =====================================================================
import pandas as pd
import matplotlib.pyplot as plt

# Files, colours and line styles
files={
    "Gz":"r_gz_Conver_Data.xlsx",
    "Txy":"r_txy_Conver_Data.xlsx",
    "Tzx":"r_tzx_Conver_Data.xlsx",
    "Tzy":"r_tzy_Conver_Data.xlsx"
}
colors={"Gz":"#0072B2","Txy":"#009E73","Tzx":"#E69F00","Tzy":"#CC79A7"}
styles={"Gz":"-","Txy":"--","Tzx":"-.","Tzy":":"}

# =====================================================================
# READ DATA AND PLOT
# =====================================================================
fig,ax=plt.subplots(figsize=(9,6))

for label,file in files.items():
    df=pd.read_excel(file)
    ax.semilogy(df["Iteration"],df["phi_d"],lw=2,
                color=colors[label],linestyle=styles[label],label=label)

# =====================================================================
# AXES AND TICKS
# =====================================================================
ax.set_xlabel("Iteration",fontsize=15,fontweight="bold")
ax.set_ylabel(r"Data misfit ($\phi_d$)",fontsize=15,fontweight="bold")
ax.tick_params(axis="both",labelsize=15,direction="out",width=1)

ax.set_ylim(10**2,10**6)
ax.set_yticks([10**2,10**3,10**4,10**5,10**6])
ax.set_yticklabels([
    r"$\mathbf{10^2}$",
    r"$\mathbf{10^3}$",
    r"$\mathbf{10^4}$",
    r"$\mathbf{10^5}$",
    r"$\mathbf{10^6}$"
],fontsize=15,fontweight="bold")

# Make all tick labels bold
for tick in ax.get_xticklabels():
    tick.set_fontweight("bold")

# =====================================================================
# GRID
# =====================================================================
ax.grid(
    True,
    which="both",
    linestyle="--",
    linewidth=.6,
    alpha=.3
)
# =====================================================================
# LEGEND
# =====================================================================
leg=ax.legend(
    loc="upper right",
    fontsize=14,
    frameon=True,
    framealpha=1
)
for text in leg.get_texts():
    text.set_fontweight("bold")
# =====================================================================
# SAVE FIGURE
# =====================================================================
fig.subplots_adjust(
    left=.10,
    right=.97,
    bottom=.10,
    top=.97
)
plt.savefig(
    "Real_Data_Convergence_Curves_All_Components.png",
    dpi=600,
    bbox_inches="tight",
    facecolor="white"
)
plt.show()
plt.close(fig)
```
